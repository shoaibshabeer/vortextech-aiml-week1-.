```python
# Age: numeric, right-skewed -> median is more robust than mean
df['Age'] = df['Age'].fillna(df['Age'].median())

# Embarked: categorical, only 2 missing -> fill with the most common port
df['Embarked'] = df['Embarked'].fillna(df['Embarked'].mode()[0])

# Cabin: missing for ~77% of rows -> too sparse to impute meaningfully,
# so convert it into a flag: do we know the cabin or not
df['HasCabin'] = df['Cabin'].notnull().astype(int)
df.drop(columns=['Cabin'], inplace=True)

df.isnull().sum()
```

```python
df['Survived'] = df['Survived'].astype('category')
df['Pclass'] = df['Pclass'].astype('category')
df['Sex'] = df['Sex'].astype('category')
df['Embarked'] = df['Embarked'].astype('category')

df.dtypes
```

```python
df.describe()
```

```python
df['Survived'].value_counts()
```

```python
df.groupby('Pclass', observed=True)['Fare'].mean().sort_values(ascending=False)
```

```python
plt.figure(figsize=(8, 5))
df['Age'].hist(bins=25, color='#4C72B0', edgecolor='white')
plt.title('Distribution of Passenger Age')
plt.xlabel('Age')
plt.ylabel('Number of Passengers')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))
survival_by_class = df.groupby('Pclass', observed=True)['Survived'].apply(lambda s: s.astype(int).mean())
sns.barplot(x=survival_by_class.index.astype(str), y=survival_by_class.values,
            hue=survival_by_class.index.astype(str), palette='viridis', legend=False)
plt.title('Survival Rate by Passenger Class')
plt.xlabel('Passenger Class')
plt.ylabel('Survival Rate')
plt.tight_layout()
plt.show()
```
